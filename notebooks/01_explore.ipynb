### Zero screenings 
A screening count of 0 likely means missing data, not that no kids were tested.

```python
cols = ["screened_for_lead_in_1999", "screened_for_lead_in_2000", "screened_for_lead_in_2001",
        "screened_for_lead_in_2002", "screened_for_lead_in_2003"]
(df[cols] == 0).sum()
```

```python
df[df["screened_for_lead_in_2003"] == 0][["community_area_number", "community_area_name", "screened_for_lead_in_2003", "percent_elevated_2003"]]
```

### 2005 outlier
The 2005 max is far above neighboring years. Checking which row it comes from.

```python
df[["percent_elevated_2004", "percent_elevated_2005", "percent_elevated_2006"]].describe()
```

```python
df.nlargest(3, "percent_elevated_2005")[["community_area_number", "community_area_name", "screened_for_lead_in_2005", "percent_elevated_2005"]]
```

The 75.3% belongs to the unknown (#99), which is already being dropped.

### Do population names match the lookup?
Any name that doesn't match exactly would drop out of the join.

```python
set(pop["community_area"]) - set(lookup["community"])
```

```python
set(lookup["community"]) - set(pop["community_area"])
```

Both empty, so all 77 names match.

## Cleaning plan

Based on this exploration:

**Lead**
- Keep community areas 1-77 (drops citywide #0 and Unknown #99)
- Convert zero screenings to missing (Riverdale 2003 and 2005)
- Reshape from wide to one row per neighborhood per year
- Average 1999-2003 into one exposure value per neighborhood

**Crime**
- Keep community areas 1-77 (drops area 0)
- Keep only violent crime codes: 01A, 02, 03, 04A, 04B

**Socioeconomic**
- Drop the citywide row
- Convert `ca` to integer
- Rename `per_capita_income_` to `per_capita_income`

**Population**
- Convert `total_population` to integer
- Attach community area numbers using the lookup table
